# Carga, extracción y transformación de datos con Python

## Importación de librerías

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from google.colab import files
```

```python
df = pd.read_csv("bank-additional-full.csv", sep=";")
df.head(10)
```

## Exploración de los datos

```python
# La forma del dataframe (filas x columna)
print(df.shape)
```

```python
# Tipos de datos en dataframe
print(df.dtypes)
```

```python
# Resumen estadístico del dataframe
df.describe(include="all")
```

```python
# Conteo de nulos o vacios
df.isna().sum()
```

## Limpieza de datos

```python
df["education"] = df["education"].fillna("Sin información")
df["education"] = df["education"].replace("unknown", "Sin información")
```

```python
df.head(10)
```

```python
df["default"] = df["default"].fillna("Sin información")
df["default"] = df["default"].replace("unknown", "Sin información")
```

```python
df.head(10)
```

## Transformación

```python
#Mes
df["fulldate"] = "01-2025-" + df["month"].astype(str)
df["fulldate"] = pd.to_datetime(df["fulldate"], format="%d-%Y-%b")
df.head(10)
```

## Visualización

```python
plt.figure(figsize=(6,3))
sns.histplot(df["age"], bins=30, kde=True, color="purple")

plt.title("Distribución por edades", fontsize=16)
plt.xlabel("Edad")
plt.ylabel("Frecuencia")
plt.axvline(df["age"].mean(), color="red")
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.boxplot(x="marital", y="age", data=df)

plt.title("Distribución de edad por estado civil", fontsize=16)
plt.xlabel("Estado civil")
plt.ylabel("Edad")
plt.show()
```
